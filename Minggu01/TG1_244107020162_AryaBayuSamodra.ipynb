3. Jelaskan kegunaan masing-masing library tersebut:
    - PyPREP: berfokus pada tahap pembersihan awal (preprocessing) khusus sinyal EEG. Library ini bekerja mengotomatisasi pendeteksian kanal sinyal yang rusak (bad channels), mereduksi gangguan listrik lingkungan, serta melakukan kalkulasi re-referensi sinyal secara obyektif sehingga data rekam otak siap dianalisis lebih lanjut.
    - SciPy: bertindak sebagai mesin komputasi matematis dan pemrosesan sinyal digital (Digital Signal Processing) umum yang menjadi fondasi analisis biosinyal. SciPy digunakan untuk membuat filter frekuensi (seperti bandpass filter untuk membuang gelombang pengganggu), menghitung Transformasi Fourier (FFT) guna menganalisis spektrum frekuensi, serta menangani kalkulasi statistik dan manipulasi matriks.
    - wandb: berperan di tahap pengembangan model kecerdasan buatan. Setelah sinyal EEG atau ECG selesai dibersihkan dan siap dimasukkan ke dalam model klasifikasi, wandb mencatat seluruh eksperimen secara real-time. Library ini merekam metrik seperti loss dan akurasi, menyimpan visualisasi grafik, serta membandingkan berbagai konfigurasi hyperparameter melalui dasbor web.
    - pyECG: dirancang khusus untuk memproses dan menganalisis sinyal elektrokardiogram (ECG). Library ini berfokus pada organ jantung dengan mendeteksi titik-titik puncak gelombang (seperti R-peak), mengukur variabilitas detak jantung (Heart Rate Variability / HRV), serta mengekstraksi fitur-fitur kardiovaskular penting dari sinyal rekam jantung.
