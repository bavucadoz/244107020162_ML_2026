# **Praktikum 1**
*Praktikum 1 ini akan membuat JST sederhana (2 layer) dengan forward pass dan backpropagation manual.*


---



**Langkah-langkah:**
1. Buat dataset sederhana (XOR).

2. Inisialisasi bobot dan bias.

3. Implementasikan forward pass.

4. Hitung error dan lakukan backpropagation.

5. Update bobot menggunakan gradient descent.

```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter
input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

# Output akhir
print("Prediksi:")
print(a2)
```

**Tugas 1:**

- Ubah jumlah neuron hidden layer menjadi 3.

- Bandingkan hasil loss dengan konfigurasi awal.

- Tambahkan fungsi aktivasi ReLU dan bandingkan hasil.

```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Fungsi aktivasi & turunannya
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(a): # a adalah output sigmoid(x)
    return a * (1 - a)

def relu(x):
    return np.maximum(0, x)

def relu_derivative(z): # z adalah input sebelum aktivasi
    return (z > 0).astype(float)

def train_xor(hidden_size=2, activation="sigmoid", epochs=10000, lr=0.1, seed=42):
    np.random.seed(seed)
    input_size = 2
    output_size = 1

    # Inisialisasi bobot (He initialization untuk ReLU agar tidak mudah dead neuron)
    if activation == "relu":
        W1 = np.random.randn(input_size, hidden_size) * np.sqrt(2. / input_size)
    else:
        W1 = np.random.randn(input_size, hidden_size)

    b1 = np.zeros((1, hidden_size))
    W2 = np.random.randn(hidden_size, output_size)
    b2 = np.zeros((1, output_size))

    for epoch in range(epochs):
        # Forward pass
        z1 = np.dot(X, W1) + b1
        if activation == "relu":
            a1 = relu(z1)
        else:
            a1 = sigmoid(z1)

        z2 = np.dot(a1, W2) + b2
        a2 = sigmoid(z2)

        # Hitung error
        error = y - a2
        loss = np.mean(np.square(error))

        # Backpropagation
        d_a2 = error * sigmoid_derivative(a2)
        d_W2 = np.dot(a1.T, d_a2)
        d_b2 = np.sum(d_a2, axis=0, keepdims=True)

        if activation == "relu":
            d_a1 = np.dot(d_a2, W2.T) * relu_derivative(z1)
        else:
            d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)

        d_W1 = np.dot(X.T, d_a1)
        d_b1 = np.sum(d_a1, axis=0, keepdims=True)

        # Update bobot
        W1 += lr * d_W1
        b1 += lr * d_b1
        W2 += lr * d_W2
        b2 += lr * d_b2

    return loss, a2

# Jalankan ketiga perbandingan
loss_awal, pred_awal = train_xor(hidden_size=2, activation="sigmoid")
loss_h3_sig, pred_h3_sig = train_xor(hidden_size=3, activation="sigmoid")
loss_h3_relu, pred_h3_relu = train_xor(hidden_size=3, activation="relu")

print(f"1. Awal (Hidden=2, Sigmoid) -> Loss Akhir: {loss_awal:.6f}")
print(f"2. Ubah (Hidden=3, Sigmoid) -> Loss Akhir: {loss_h3_sig:.6f}")
print(f"3. Ubah (Hidden=3, ReLU)    -> Loss Akhir: {loss_h3_relu:.6f}\n")

print("Prediksi Akhir (Hidden=3, ReLU):")
print(np.round(pred_h3_relu, 4))
```
