# **Tugas Praktikum**
*Pada tugas praktikum ini kita akan menggunakan lJST untuk melakukan klasifikasi angka tulisan tangan (MNIST).*


---



**Langkah-langkah:**
1. Load dataset MNIST dari Keras.

2. Bangun model dengan 2 hidden layer.

3. Latih model dan evaluasi akurasi.

```python
import time
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 -> 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 2. Bangun model JST
model = Sequential([
    Flatten(input_shape=(28, 28)),        # Ubah gambar 28x28 menjadi vektor 784
    Dense(128, activation='relu'),        # Hidden layer 1
    Dense(64, activation='relu'),         # Hidden layer 2
    Dense(10, activation='softmax')       # Output layer (10 kelas)
])

# 3. Kompilasi model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# 4. Latih model & ukur waktu pelatihan
start_time = time.time()
model.fit(X_train, y_train, epochs=10, batch_size=128, validation_split=0.1)
waktu_latih = time.time() - start_time

# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\nWaktu Pelatihan     : {waktu_latih:.2f} detik")
print(f"Loss pada data uji  : {loss:.4f}")
print(f"Akurasi pada data uji: {acc:.4f} ({acc*100:.2f}%)")
```



---

**Melakukan percobaan dengan beberapa parameter lain:**

- Ubah jumlah neuron di hidden layer (misal: 256 dan 128).

- Tambahkan satu hidden layer lagi.

- Bandingkan akurasi dan waktu pelatihan.

- Eksperimen dengan fungsi aktivasi Sigmoid vs ReLU.

```python
import numpy as np

# Load & Preprocessing Data
(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train, X_test = X_train / 255.0, X_test / 255.0
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

def uji_model_mnist(nama_eksperimen, layers_config, aktivasi='relu', epochs=10):
    tf.random.set_seed(42)
    np.random.seed(42)

    model = Sequential()
    model.add(Flatten(input_shape=(28, 28)))

    for units in layers_config:
        model.add(Dense(units, activation=aktivasi))

    model.add(Dense(10, activation='softmax'))

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    mulai = time.time()
    model.fit(X_train, y_train, epochs=epochs, batch_size=128, verbose=0)
    durasi = time.time() - mulai

    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    return {
        "Skenario": nama_eksperimen,
        "Arsitektur": str(layers_config),
        "Aktivasi": aktivasi,
        "Waktu (detik)": durasi,
        "Test Loss": loss,
        "Akurasi (%)": acc * 100
    }

# Jalankan 4 skenario eksperimen
daftar_uji = [
    uji_model_mnist("1. Dasar (2 Hidden Layer)", [128, 64], aktivasi='relu'),
    uji_model_mnist("2. Neuron Diperbesar", [256, 128], aktivasi='relu'),
    uji_model_mnist("3. Tambah 1 Hidden Layer (3 Layer)", [256, 128, 64], aktivasi='relu'),
    uji_model_mnist("4. Aktivasi Sigmoid", [256, 128], aktivasi='sigmoid')
]

print(f"{'Skenario':<35} | {'Arsitektur':<15} | {'Aktivasi':<8} | {'Waktu':<9} | {'Loss':<8} | {'Akurasi'}")
print("-" * 98)
for h in daftar_uji:
    print(f"{h['Skenario']:<35} | {h['Arsitektur']:<15} | {h['Aktivasi']:<8} | {h['Waktu (detik)']:>6.2f} dtk | {h['Test Loss']:<8.4f} | {h['Akurasi (%)']:.2f}%")
```
